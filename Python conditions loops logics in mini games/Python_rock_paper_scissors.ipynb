# Rock, Paper, Scissors Game

```python
def Rock_paper_scissors_game():
    Name = input('Enter your name').capitalize()
    print("Hi",Name,"welcome to the Game of Rock,Paper,Scissors.")
    print("""rules: player has given with chance to choose between Rock, Paper & Scissors
           Values - Rock > Scissors, Scissors > Paper, Paper > Rock.  Choose wisely""")
    import random
    game = ['Rock','Paper','Scissors']
    
    
    while True:
        Player = input("choose Between (Rock, Paper & Scissors)").capitalize()
        bot = random.choice(game)
        
        if Player not in game:
                print("Use only between Rock, Paper & Scissors")
                continue
    
        print('bot choses',bot)
    
        if Player == bot:
            print('Match Draw')
        elif (Player == 'Rock' and bot == 'Scissors') or (Player == 'Paper' and bot == 'Rock') or (Player == 'Scissors' and bot == 'Paper'):
            print("you wins, I know you will")
    
        else:
            print("bot wins, better luck next time")
    
        play_again = input('Want to play again? (yes/no)').lower()
    
        if play_again != 'yes':
            print('Game over')
            print('Thanks for the visit',Name)
            break
            
```

```python
Rock_paper_scissors_game()
```
