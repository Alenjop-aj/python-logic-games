# Word Guessing Game (Mini Hangman)

```python
def Missing_word_guessing_game():
    import random
    print("Welcome to Mystery word find it dood?")
    tricky_List = ['pipeline','automation','portfolio','research','statistics','freelancing','developers','generative']
    mystery_word = random.choice(tricky_List)
    tried_letters = ''
    chances = 6
    while chances > 0:
        hidden_letters = 0
        for letters in mystery_word:
           if letters in tried_letters:
               print(letters, end=' ')
           else:
               print('_', end='_')
               hidden_letters +=1
        if hidden_letters == 0:
                print('wow you guessed the Mystery word ***{}*** correctly'.format(mystery_word))
                break
            
        player_guess = input('\nguess the letters: ').lower()
        
        if player_guess in tried_letters:
            print('Guessed already, try another letter')
            continue
        if len(player_guess) != 1:
            print('Guess only one letter per chance')
            continue
        tried_letters += player_guess   
        if player_guess not in mystery_word:
            chances -= 1
            print('Oops wrong letter, you still have {} more guesses'.format(chances))
        if chances == 0:
            print('game over, the correct mystery word was ***{}***'.format(mystery_word))
        if player_guess in mystery_word:
             print('correct guess')
        if chances <= 0:
            print('All your attempts are over, Kindly tryagain')
    
```

```python
Missing_word_guessing_game()
```
